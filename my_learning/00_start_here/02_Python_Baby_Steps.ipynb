## 自动批改练习

每道题先读题、再写代码、最后运行同一格。看到 `[通过]` 才算完成；`[未通过]` 会说明哪里不对。
参考答案默认折叠，建议独立尝试至少 10 分钟后再看。


````python
# 自动检查工具：每道练习末尾都会调用 check(test_xxx)。
# 未完成或有错时只打印提示，不会中断 Run All。
def check(test):
    try:
        test()
    except NotImplementedError:
        print(f"[未完成] {test.__name__}：把 raise NotImplementedError 换成你的代码，再运行本格。")
    except AssertionError as error:
        print(f"[未通过] {test.__name__}：{error}")
    except Exception as error:  # 代码本身报错，也给出可读提示
        print(f"[代码报错] {test.__name__}：{type(error).__name__}: {error}")
    else:
        print(f"[通过] {test.__name__}")
````

### 练习 3：P03：自己写累计循环

用for实现奖励求和，空列表结果为0。

<details><summary>提示（卡住 10 分钟再看）</summary>

先在纸上填写每轮reward和total；测试不会判断是否用了for，请自行遵守练习要求。

</details>


````python
def sum_rewards(rewards):
    raise NotImplementedError


# ---- 自动检查（不用修改） ----
def test_sum_rewards():
    assert sum_rewards([1, 0, 2]) == 3, 'total初始化放在循环外'
    assert sum_rewards([]) == 0, '空列表应返回初始总和'
    assert sum_rewards([-2, 3, 4]) == 5, '不能只返回最后一项'

check(test_sum_rewards)
````

### 练习 4：P04：折扣回报

按从左到右顺序，第一项权重1，后一项权重乘gamma。空列表返回0。

<details><summary>提示（卡住 10 分钟再看）</summary>

维护weight=1，每处理一项后才乘gamma。

</details>


````python
def discounted_return(rewards, gamma):
    raise NotImplementedError


# ---- 自动检查（不用修改） ----
def test_discounted_return():
    assert abs(discounted_return([1, 2, 3], .5) - 2.75) < 1e-9, '逐项算1+1+0.75'
    assert discounted_return([2, 4], 0) == 2, 'gamma=0只留第一项'
    assert discounted_return([2, 4], 1) == 6, 'gamma=1是普通总和'
    assert discounted_return([], .9) == 0, '不要除以列表长度'

check(test_discounted_return)
````

<details><summary>参考答案（先独立完成，再对照）</summary>

```python
def discounted_return(rewards, gamma):
    total = 0.0
    weight = 1.0
    for reward in rewards:
        total += weight * reward
        weight *= gamma
    return total
```

### 练习 5：P04：return与默认参数

写average_reward，空列表返回默认值default（默认0），非空返回平均值。

<details><summary>提示（卡住 10 分钟再看）</summary>

空列表不能做除法；先判断后返回。

</details>


````python
def average_reward(rewards, default=0):
    raise NotImplementedError


# ---- 自动检查（不用修改） ----
def test_average_reward():
    assert average_reward([1, 2]) == 1.5, '平均用/，不是整除//'
    assert average_reward([]) == 0, '先处理空列表'
    assert average_reward([], default=-1) == -1, '调用参数覆盖默认值'

check(test_average_reward)
````

<details><summary>参考答案（先独立完成，再对照）</summary>

```python
def average_reward(rewards, default=0):
    if not rewards:
        return default
    return sum(rewards) / len(rewards)
```

### 练习 6：P05：对象各有自己的数据

创建Counter，初始steps=0；每次step()累加1并返回新的steps。

<details><summary>提示（卡住 10 分钟再看）</summary>

self代表当前对象；不是一个所有对象共用的全局变量。

</details>


````python
class Counter:
    def __init__(self):
        raise NotImplementedError

    def step(self):
        raise NotImplementedError


# ---- 自动检查（不用修改） ----
def test_counter():
    first = Counter()
    second = Counter()
    assert first.steps == 0, '初始化实例属性self.steps'
    assert first.step() == 1, '方法需要return新计数'
    assert first.step() == 2, '调用后保留状态'
    assert second.steps == 0, '两个实例的状态应独立'

check(test_counter)
````

<details><summary>参考答案（先独立完成，再对照）</summary>

```python

    def __init__(self):
        self.steps = 0

    def step(self):
        self.steps += 1
        return self.steps
```
