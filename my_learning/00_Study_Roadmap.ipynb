## 4. 什么时候算“掌握”

一个知识点至少经过四层：

1. **识别：**看到公式知道各符号含义。
2. **解释：**不用公式说清它解决的问题。
3. **实现：**从空函数写出核心更新，并通过小测试。
4. **诊断：**曲线异常时能提出有顺序的排查假设。

只完成前两层叫“理解”，四层都完成才叫“掌握”。用 [LEARNING_CHECKLIST.md](LEARNING_CHECKLIST.md) 持续打卡。
